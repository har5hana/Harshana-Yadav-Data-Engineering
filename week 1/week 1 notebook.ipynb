```python
import pandas as pd
import os
from glob import glob

folder_path = r"C:\Users\HP\Documents\celebal\celebal week 1 data"

csv_files = glob(os.path.join(folder_path, "*.csv"))

print("CSV files found:", len(csv_files))

df_list = []

for file in csv_files:
    print("Loading:", os.path.basename(file))
    temp_df = pd.read_csv(file)
    df_list.append(temp_df)

df = pd.concat(df_list, ignore_index=True)

print("\nCombined Shape:", df.shape)

df.head()
```

```python
#handling missing values
missing_values= df.isnull().sum()
print(missing_values)
```

```python
#totaling all missing values
print("total missing values : ",df.isnull().sum().sum())
```

```python
#filling numeric missing values
numeric_cols= df.select_dtypes(include = ['number']).columns
for col in numeric_cols:
    df[col] =df[col].fillna(df[col].mean())
```

```python
#filling text missing values
text_cols=df.select_dtypes(include=['object']).columns
for col in text_cols:
    df[col]=df[col].fillna('unknown')
print(df.isnull().sum())
```

```python
print(df.columns.tolist())
```

```python
#basic operations
#select columns
selected_columns =df[['title','rating','final_price']]
selected_columns.head()
```

```python
#products with rating greater than 4
high_rated_products=df[df['rating']>4]
high_rated_products.head()
```

```python
#products with discount greater than 50%
high_discount_products=df[df['discount']>50]
high_discount_products.head()
```

```python
df['final_price'] = (
    df['final_price']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
)
df['final_price'] = pd.to_numeric(df['final_price'], errors='coerce')
```

```python
print(df['final_price'].dtype)
```

```python
#products worth more than 1000
expensive_products=df[df['final_price']>1000]
expensive_products.head()
```

```python
print("duplicates befpre cleaning : ",df.duplicated().sum())
```

```python
df=df.drop_duplicates()
print("duplicates after cleaning : ",df.duplicated().sum())
```

```python
df = df.copy()
df['quantity']=1
df['total_amount']=df['final_price']*df['quantity']
```

```python
df.to_csv('cleaned_dataset.csv',index=False)
print("cleaned dataset saved")
```
